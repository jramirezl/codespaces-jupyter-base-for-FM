## Preguntas

### 1. (5 puntos) Normas.

Demuestre la siguiente identidad:

\[
\text{rms}^2_w(x) = \text{avg}^2_w(x) + \text{std}^2_w(x)
\]

donde \( w_1, w_2, \ldots, w_n \) son pesos positivos tales que la suma total de los pesos es

\[
W = \sum_{i=1}^n w_i > 0,
\]

y donde tenemos las definiciones:

\[
\text{avg}_w(x) = \frac{1}{W} \sum_{i=1}^n w_i x_i,
\]

\[
\text{std}^2_w(x) = \frac{1}{W} \sum_{i=1}^n w_i \left(x_i - \text{avg}_w(x)\right)^2,
\]

\[
\text{rms}_w(x) = \sqrt{ \frac{1}{W} \sum_{i=1}^n w_i x_i^2 }.
\]
